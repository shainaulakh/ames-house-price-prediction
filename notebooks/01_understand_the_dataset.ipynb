# Part 1 · Understanding Ames house sales

**Question:** Before a home sells, can we use its characteristics to estimate its final sale price? A sale is one example. Its recorded price is the answer a future model will learn to estimate. Predicting a dollar amount is a regression task.

## Load the sales

OpenIntro distributes these historical Ames Assessor’s Office records as a CSV. Pandas reads the header as column names and stores the sales in a table called a DataFrame.

```python
from pathlib import Path
import pandas as pd

# VS Code can start the kernel in the project root or the notebooks folder.
root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
sales = pd.read_csv(root / "data" / "ames.csv")
rows, columns = sales.shape
print(f"{rows:,} recorded sales and {columns} columns")
```

## Identify the clues and the answer

| Column | Meaning | Role |
|---|---|---|
| area | Above-ground living area in square feet | Numeric feature |
| Neighborhood | Neighbourhood code | Categorical feature |
| Year.Built | Original construction year | Numeric feature |
| price | Recorded sale price in US dollars | Target |

Column names here match OpenIntro’s CSV, including capitalization. Selecting three features keeps the first inspection manageable.

```python
features = ["area", "Neighborhood", "Year.Built"]
target = "price"
preview = sales[features + [target]]
print(preview.head(3).to_string(index=False))
```

```python
print(preview.dtypes.to_string())
```

Read the first row from left to right. The first three entries describe the home. The final entry is the amount it sold for. Pandas stores the neighbourhood codes as text and the other three columns as integers in this file.
